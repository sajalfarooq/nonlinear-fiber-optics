# Chirp-controlled compression

The compression metric is the minimum RMS intensity width relative to input. The best sampled chirp can lie at a scan boundary; it is not automatically a global optimum.

All measurements here are synthetic. This notebook runs a small independent calculation and then inspects a retained study figure. It does not launch full training or overwrite the saved scientific results.

```python

from pathlib import Path
import json
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, Image
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "pyproject.toml").exists() and (p / "src/optics_portfolio").exists())
%matplotlib inline

PROJECT = ROOT
```

## Calculate and inspect

Read the calculation before executing. Compare numerical values with the stated physical assumptions.

```python
data=np.load(PROJECT/"results/study_data.npz")
best=int(np.argmin(data["minimum_width_ratio"]))
print("Best sampled chirp:",data["chirps"][best])
print("Minimum width ratio:",data["minimum_width_ratio"][best])
print("Distance:",data["compression_distance"][best])
plt.plot(data["chirps"],data["minimum_width_ratio"],"o-")
plt.xlabel("Input chirp C"); plt.ylabel("Minimum RMS width / initial width"); plt.show()
```
